```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._marks = 0
        self.__grade = "F"
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self._marks = marks

            if marks >= 80:
                self.__grade = "A"
            elif marks >= 60:
                self.__grade = "B"
            elif marks >= 40:
                self.__grade = "C"
            else:
                self.__grade = "F"

            print("Marks updated successfully.")
        else:
            print("Invalid marks! Enter marks between 0 and 100.")

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


# Creating objects
s1 = Student("Ali", 85)
s2 = Student("Ahmed", 55)

print("\nStudent:", s1.name)
print("Marks:", s1.get_marks())
print("Grade:", s1.get_grade())

print("\nStudent:", s2.name)
print("Marks:", s2.get_marks())
print("Grade:", s2.get_grade())

# Valid update
s1.set_marks(95)

# Invalid update
s2.set_marks(120)

# Direct access to private attribute
try:
    print(s1.__grade)
except AttributeError:
    print("Cannot access private grade directly!")
```

```python
class BankAccount:
    def __init__(self, name, balance, pin):
        self.name = name
        self.__balance = balance
        self.__pin = pin

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful!")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Incorrect PIN!")
        elif amount <= 0:
            print("Invalid withdrawal amount!")
        elif amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print("Withdrawal successful!")

    def get_balance(self, pin):
        if pin == self.__pin:
            print("Current Balance:", self.__balance)
        else:
            print("Incorrect PIN! Access denied.")


# Creating account
account = BankAccount("Ali", 10000, 1234)

# Successful deposit
account.deposit(5000)

# Successful withdrawal
account.withdraw(3000, 1234)

# Rejected operation 1
account.withdraw(20000, 1234)

# Rejected operation 2
account.deposit(-500)

# Display balance
account.get_balance(1234)

# Incorrect PIN
account.get_balance(1111)
```

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_details(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id)
        self.language = language

    def display_details(self):
        print("\nDeveloper Details")
        super().display_details()
        print("Programming Language:", self.language)


class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def display_details(self):
        print("\nManager Details")
        super().display_details()
        print("Team Size:", self.team_size)


# Creating objects
d1 = Developer("Ali", 101, "Python")
m1 = Manager("Ahmed", 102, 10)

d1.display_details()
m1.display_details()
```

```python
class Vehicle:
    def __init__(self, brand, model, rate):
        self.brand = brand
        self.model = model
        self.rate = rate

    def calculate_rent(self, days):
        if days <= 0:
            print("Invalid rental duration!")
            return None

        return self.rate * days


class Car(Vehicle):
    def calculate_rent(self, days):
        if days <= 0:
            print("Invalid rental duration!")
            return None

        rent = super().calculate_rent(days)
        return rent + 1000


class Bike(Vehicle):
    def calculate_rent(self, days):
        if days <= 0:
            print("Invalid rental duration!")
            return None

        rent = super().calculate_rent(days)
        return rent - (rent * 0.10)


# Creating objects
car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD70", 2000)

# Three-day rental
car_rent = car.calculate_rent(3)
bike_rent = bike.calculate_rent(3)

print("Car Rental:", car_rent)
print("Bike Rental:", bike_rent)
```

```python
class EmailNotification:
    def send(self, message):
        print("Email Notification:", message)


class SMSNotification:
    def send(self, message):
        print("SMS Notification:", message)


class PushNotification:
    def send(self, message):
        print("Push Notification:", message)


class WhatsAppNotification:
    def send(self, message):
        print("WhatsApp Notification:", message)


def notify_all(notifications, message):
    for notification in notifications:
        notification.send(message)


# Creating objects
notifications = [
    EmailNotification(),
    SMSNotification(),
    PushNotification(),
    WhatsAppNotification()
]

notify_all(notifications, "Your order has been confirmed!")
```

```python
class RegularCustomer:
    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount - (amount * 0.05)


class PremiumCustomer:
    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount - (amount * 0.10)


class VIPCustomer:
    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount - (amount * 0.20)


class GoldCustomer:
    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount - (amount * 0.15)


def print_receipt(customer, amount):
    price = customer.final_price(amount)

    if price is not None:
        print("Original Amount: Rs.", amount)
        print("Final Price: Rs.", price)
        print("----------------------")


# Creating objects
regular = RegularCustomer()
premium = PremiumCustomer()
vip = VIPCustomer()
gold = GoldCustomer()

# Printing receipts
print("Regular Customer")
print_receipt(regular, 10000)

print("Premium Customer")
print_receipt(premium, 10000)

print("VIP Customer")
print_receipt(vip, 10000)

print("Gold Customer")
print_receipt(gold, 10000)
```

```python
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14159 * self.radius


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


# Creating objects
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle Area:", circle.area())
print("Circle Perimeter:", circle.perimeter())

print("\nRectangle Area:", rectangle.area())
print("Rectangle Perimeter:", rectangle.perimeter())


# Trying to create abstract class object
try:
    shape = Shape()
except TypeError:
    print("\nCannot create Shape object directly!")
```

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Card Payment - Pay: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Card Payment - Refund: Rs.", amount)


class WalletPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Wallet Payment - Pay: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Wallet Payment - Refund: Rs.", amount)


class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Bank Transfer - Pay: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Bank Transfer - Refund: Rs.", amount)


def process_transaction(method, amount):
    method.pay(amount)


# Creating objects
card = CardPayment()
wallet = WalletPayment()
bank = BankTransferPayment()

# Processing payments
process_transaction(card, 5000)
process_transaction(wallet, 3000)
process_transaction(bank, 7000)

print("\nRefund Examples")

card.refund(1000)
wallet.refund(500)
bank.refund(2000)

# Invalid transaction
print("\nInvalid Transaction")
process_transaction(card, -500)
```
