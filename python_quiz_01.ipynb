```python
def analyze_transactions(transactions, customers):
    seen_ids = set()
    spending = {}
    categories = {}

    for customer in customers:
        spending[customer] = 0
        categories[customer] = {}

    for t in transactions:
        tid = t["id"]

        if tid in seen_ids:
            raise ValueError("Duplicate transaction ID")

        seen_ids.add(tid)

        customer = t["customer"]

        if customer not in spending:
            spending[customer] = 0
            categories[customer] = {}

        if t["status"] == "success":
            amount = t["amount"]
            category = t["category"]

            spending[customer] += amount

            categories[customer][category] = (
                categories[customer].get(category, 0) + amount
            )

    report = []

    for customer, total in spending.items():
        top_category = None

        if categories[customer]:
            top_category = max(
                categories[customer],
                key=categories[customer].get
            )

        report.append({
            "customer": customer,
            "total_spending": total,
            "top_category": top_category
        })

    report.sort(key=lambda x: (-x["total_spending"], x["customer"]))

    for rank, item in enumerate(report, start=1):
        item["rank"] = rank

    return report


transactions = [
    {"id": 1, "customer": "faizan", "category": "Food", "amount": 500, "status": "success"},
    {"id": 2, "customer": "faizan", "category": "Travel", "amount": 200, "status": "failed"},
    {"id": 3, "customer": "usama", "category": "Food", "amount": 700, "status": "success"}
]

customers = ["faizan", "usama", "Ahmed"]

print(analyze_transactions(transactions, customers))
```

```python
def execution_order(tasks, dependencies):
    graph = {task: [] for task in tasks}
    indegree = {task: 0 for task in tasks}

    for a, b in dependencies:
        if a not in graph or b not in graph:
            raise ValueError("Unknown task name")

        if b not in graph[a]:
            graph[a].append(b)
            indegree[b] += 1

    available = sorted(
        [task for task in tasks if indegree[task] == 0]
    )

    result = []

    while available:
        current = available.pop(0)
        result.append(current)

        for neighbor in sorted(graph[current]):
            indegree[neighbor] -= 1

            if indegree[neighbor] == 0:
                available.append(neighbor)
                available.sort()

    if len(result) != len(tasks):
        raise ValueError("Cycle detected in dependencies")

    return result


tasks = ["build", "test", "deploy"]
dependencies = [("build", "test"), ("test", "deploy")]

print(execution_order(tasks, dependencies))
```

```python
import time
import functools


def ttl_cache(seconds):

    def decorator(function):  
        cache = {}
        stats = {"hits": 0, "misses": 0}

        @functools.wraps(function)
        def wrapper(*args, **kwargs):
            try:
                key = (args, tuple(sorted(kwargs.items())))
                hash(key)
            except TypeError:
                raise TypeError("Cache arguments must be hashable")

            current_time = time.monotonic()

            if key in cache:
                result, timestamp = cache[key]

                if current_time - timestamp < seconds:
                    stats["hits"] += 1
                    return result

            stats["misses"] += 1
            result = function(*args, **kwargs)
            cache[key] = (result, current_time)

            return result

        wrapper.cache_info = lambda: stats.copy()
        return wrapper

    return decorator


@ttl_cache(seconds=5)
def add(x, y):
    print("Calculating...")
    return x + y


print(add(x=2, y=3))
print(add(y=3, x=2))
print(add.cache_info())
```

```python
from collections import defaultdict
from datetime import datetime

malformed_count = 0


def read_errors(path):
    global malformed_count
    malformed_count = 0

    with open(path, "r") as file:
        for line_number, line in enumerate(file, start=1):
            parts = line.strip().split("|", 2)

            if len(parts) != 3:
                malformed_count += 1
                continue

            timestamp = parts[0].strip()
            level = parts[1].strip()
            message = parts[2].strip()

            try:
                datetime.strptime(timestamp, "%Y-%m-%d %H:%M:%S")
            except ValueError:
                malformed_count += 1
                continue

            if level == "ERROR":
                yield line_number, timestamp, message


def summarize_errors(path):
    summary = defaultdict(int)

    for number, timestamp, message in read_errors(path):
        date = timestamp.split()[0]
        summary[date] += 1

    return dict(summary)


with open("app.log", "w") as file:
    file.write("2026-11-07 10:30:00 | ERROR | Connection lost\n")
    file.write("2026-11-07 10:35:00 | INFO | Application started\n")
    file.write("2026-11-07 11:00:00 | ERROR | Database failed\n")
    file.write("Invalid line\n")


for error in read_errors("app.log"):
    print(error)

print("Summary:", summarize_errors("app.log"))
print("Malformed lines:", malformed_count)
```

```python
import csv


def reconcile_inventory():
    inventory = {}

    with open("inventory.csv", "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            sku = row["sku"]

            if sku in inventory:
                raise ValueError("Duplicate SKU")

            quantity = int(row["quantity"])

            if quantity < 0:
                raise ValueError("Invalid stock")

            inventory[sku] = {
                "name": row["name"],
                "quantity": quantity
            }

    rejected = []

    with open("movements.csv", "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            sku = row["sku"]
            movement = row["type"].strip().upper()

            try:
                units = int(row["units"])

                if sku not in inventory:
                    raise ValueError("Unknown SKU")

                if units <= 0:
                    raise ValueError("Invalid units")

                if movement == "IN":
                    inventory[sku]["quantity"] += units

                elif movement == "OUT":
                    if inventory[sku]["quantity"] < units:
                        raise ValueError("Insufficient stock")

                    inventory[sku]["quantity"] -= units

                else:
                    raise ValueError("Invalid movement type")

            except ValueError as error:
                row["reason"] = str(error)
                rejected.append(row)

    with open("updated_inventory.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["sku", "name", "quantity"])

        for sku, data in inventory.items():
            writer.writerow([sku, data["name"], data["quantity"]])

    with open("rejected_movements.csv", "w", newline="") as file:
        fields = ["sku", "type", "units", "reason"]
        writer = csv.DictWriter(file, fieldnames=fields)
        writer.writeheader()
        writer.writerows(rejected)

    print("Inventory updated successfully")
    print("Rejected movements:", rejected)


reconcile_inventory()
```

```python
import csv


def calculate_grade(average):
    if average >= 85:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 55:
        return "C"
    elif average >= 40:
        return "D"
    else:
        return "F"


def generate_report():
    students = {}
    scores = {}

    with open("students.csv", "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            sid = row["student_id"]

            if sid in students:
                raise ValueError("Duplicate student ID")

            students[sid] = row["name"]
            scores[sid] = []

    seen = set()

    with open("scores.csv", "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            sid = row["student_id"]
            subject = row["subject"]
            marks = float(row["marks"])

            if sid not in students:
                raise ValueError("Unknown student ID")

            if not 0 <= marks <= 100:
                raise ValueError("Invalid marks")

            key = (sid, subject)

            if key in seen:
                raise ValueError("Duplicate subject entry")

            seen.add(key)
            scores[sid].append(marks)

    report = []

    for sid, name in students.items():
        marks = scores[sid]
        average = sum(marks) / len(marks) if marks else 0

        report.append({
            "student_id": sid,
            "name": name,
            "average": round(average, 2),
            "grade": calculate_grade(average)
        })

    report.sort(key=lambda x: (-x["average"], x["student_id"]))

    for rank, student in enumerate(report, start=1):
        student["rank"] = rank

    with open("report.csv", "w", newline="") as file:
        fields = ["student_id", "name", "average", "grade", "rank"]
        writer = csv.DictWriter(file, fieldnames=fields)
        writer.writeheader()
        writer.writerows(report)

    return report


print(generate_report())
```

```python
class ReversibleRange:

    def __init__(self, start, stop, step=1):
        if step == 0:
            raise ValueError("Step cannot be zero")

        self.start = start
        self.stop = stop
        self.step = step

    def __len__(self):
        start = self.start
        stop = self.stop
        step = self.step

        if step > 0:
            return max(0, (stop - start + step - 1) // step)

        return max(0, (start - stop - step - 1) // (-step))

    def __iter__(self):
        current = self.start

        if self.step > 0:
            while current < self.stop:
                yield current
                current += self.step

        else:
            while current > self.stop:
                yield current
                current += self.step

    def __reversed__(self):
        count = len(self)

        if count == 0:
            return

        current = self.start + (count - 1) * self.step

        for _ in range(count):
            yield current
            current -= self.step

    def __contains__(self, value):
        if not isinstance(value, int):
            return False

        difference = value - self.start

        if difference % self.step != 0:
            return False

        index = difference // self.step

        return 0 <= index < len(self)


r = ReversibleRange(2, 11, 3)

print(list(r))
print(list(reversed(r)))
print(len(r))
print(5 in r)
```

```python
from datetime import datetime


class BookingScheduler:

    def __init__(self):
        self.bookings = []
        self.ids = set()

    def add_booking(self, room, start, end, booking_id):
        if booking_id in self.ids:
            return "Rejected: Duplicate booking ID"

        if end <= start:
            return "Rejected: Invalid time range"

        for booking in self.bookings:
            if booking["room"] == room:
                existing_start = booking["start"]
                existing_end = booking["end"]

                if start < existing_end and end > existing_start:
                    return "Rejected: Booking conflict"

        self.bookings.append({
            "room": room,
            "start": start,
            "end": end,
            "booking_id": booking_id
        })

        self.ids.add(booking_id)

        self.bookings.sort(key=lambda x: (x["room"], x["start"]))

        return "Booking accepted"

    def show_bookings(self):
        for booking in self.bookings:
            print(booking)


scheduler = BookingScheduler()

d = datetime.strptime

print(scheduler.add_booking(
    "A", d("09:00", "%H:%M"),
    d("10:00", "%H:%M"), 1
))

print(scheduler.add_booking(
    "A", d("10:00", "%H:%M"),
    d("11:00", "%H:%M"), 2
))

print(scheduler.add_booking(
    "A", d("09:30", "%H:%M"),
    d("10:30", "%H:%M"), 3
))

scheduler.show_bookings()
```

```python
import json
import os

FILE = "contacts.json"


def load_contacts():
    if not os.path.exists(FILE):
        return []

    try:
        with open(FILE, "r") as file:
            data = json.load(file)

            if not isinstance(data, list):
                raise ValueError("Invalid JSON structure")

            return data

    except json.JSONDecodeError:
        raise ValueError("Contact file is corrupted")


def save_contacts(contacts):
    temp_file = FILE + ".tmp"

    with open(temp_file, "w") as file:
        json.dump(contacts, file, indent=4)

    os.replace(temp_file, FILE)


def add_contact(contacts):
    contact_id = input("Enter ID: ")

    if any(c["id"] == contact_id for c in contacts):
        print("ID already exists")
        return

    name = input("Enter name: ")
    phone = input("Enter phone: ")
    email = input("Enter email: ")

    if not name.strip() or not phone.strip() or not email.strip():
        print("All fields are required")
        return

    contacts.append({
        "id": contact_id,
        "name": name,
        "phone": phone,
        "email": email
    })

    save_contacts(contacts)
    print("Contact added")


def search_contact(contacts):
    name = input("Enter name to search: ").lower()
    found = False

    for contact in contacts:
        if name in contact["name"].lower():
            print(contact)
            found = True

    if not found:
        print("Contact not found")


def update_contact(contacts):
    contact_id = input("Enter ID: ")

    for contact in contacts:
        if contact["id"] == contact_id:
            contact["name"] = input("New name: ")
            contact["phone"] = input("New phone: ")
            contact["email"] = input("New email: ")

            save_contacts(contacts)
            print("Contact updated")
            return

    print("Contact not found")


def delete_contact(contacts):
    contact_id = input("Enter ID: ")

    for contact in contacts:
        if contact["id"] == contact_id:
            contacts.remove(contact)
            save_contacts(contacts)
            print("Contact deleted")
            return

    print("Contact not found")


def list_contacts(contacts):
    if not contacts:
        print("No contacts available")
        return

    for contact in contacts:
        print(contact)


def main():
    try:
        contacts = load_contacts()
    except (ValueError, OSError) as error:
        print(error)
        return

    while True:
        print("\n===== CONTACT MANAGER =====")
        print("1. Add Contact")
        print("2. Search Contact")
        print("3. Update Contact")
        print("4. Delete Contact")
        print("5. List Contacts")
        print("6. Exit")

        choice = input("Enter choice: ")

        try:
            if choice == "1":
                add_contact(contacts)

            elif choice == "2":
                search_contact(contacts)

            elif choice == "3":
                update_contact(contacts)

            elif choice == "4":
                delete_contact(contacts)

            elif choice == "5":
                list_contacts(contacts)

            elif choice == "6":
                print("Goodbye!")
                break

            else:
                print("Invalid choice")

        except (OSError, ValueError) as error:
            print("Error:", error)


if __name__ == "__main__":
    main()
```

```python
import json
import uuid
import os


class Product:

    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price


class Inventory:

    def __init__(self, filename):
        with open(filename, "r") as file:
            self.data = json.load(file)

    def get_product(self, product_id):
        if product_id not in self.data:
            raise ValueError("Product does not exist")

        return self.data[product_id]

    def save(self):
        with open("catalog.json.tmp", "w") as file:
            json.dump(self.data, file, indent=4)

        os.replace("catalog.json.tmp", "catalog.json")


class Cart:

    def __init__(self):
        self.items = {}

    def add_item(self, product_id, quantity):
        if not isinstance(quantity, int) or quantity <= 0:
            raise ValueError("Invalid quantity")

        self.items[product_id] = (
            self.items.get(product_id, 0) + quantity
        )


class Order:

    def __init__(self, cart, inventory, discount=0):
        if not 0 <= discount <= 100:
            raise ValueError("Invalid discount")

        self.order_id = str(uuid.uuid4())
        self.cart = cart
        self.inventory = inventory
        self.discount = discount
        self.total = 0

    def calculate_total(self):
        total = 0

        for product_id, quantity in self.cart.items.items():
            product = self.inventory.get_product(product_id)
            total += product["price"] * quantity

        return total - (total * self.discount / 100)

    def place_order(self):
        for product_id, quantity in self.cart.items.items():
            product = self.inventory.get_product(product_id)

            if quantity > product["stock"]:
                raise ValueError("Out of stock")

        self.total = self.calculate_total()

        old_data = json.loads(json.dumps(self.inventory.data))

        try:
            for product_id, quantity in self.cart.items.items():
                self.inventory.data[product_id]["stock"] -= quantity

            self.inventory.save()

            order_data = {
                "order_id": self.order_id,
                "items": self.cart.items,
                "discount": self.discount,
                "total": self.total
            }

            try:
                with open("orders.json", "r") as file:
                    orders = json.load(file)

            except FileNotFoundError:
                orders = []

            orders.append(order_data)

            with open("orders.json.tmp", "w") as file:
                json.dump(orders, file, indent=4)

            os.replace("orders.json.tmp", "orders.json")

            print("Order placed successfully")
            return order_data

        except Exception:
            self.inventory.data = old_data
            self.inventory.save()

            if os.path.exists("orders.json.tmp"):
                os.remove("orders.json.tmp")

            raise


# Sample catalog

with open("catalog.json", "w") as file:
    json.dump({
        "P1": {
            "name": "Laptop",
            "price": 100,
            "stock": 5
        }
    }, file, indent=4)


inventory = Inventory("catalog.json")
cart = Cart()

cart.add_item("P1", 2)

order = Order(cart, inventory, discount=10)

print("Total:", order.calculate_total())

result = order.place_order()

print(result)
print("Remaining stock:", inventory.data["P1"]["stock"])


# UNIT TESTS

import unittest


class TestOrderSystem(unittest.TestCase):

    def setUp(self):
        with open("test_catalog.json", "w") as file:
            json.dump({
                "P1": {
                    "name": "Laptop",
                    "price": 100,
                    "stock": 5
                }
            }, file)

        self.inventory = Inventory("test_catalog.json")
        self.cart = Cart()

    def test_total(self):
        self.cart.add_item("P1", 2)
        order = Order(self.cart, self.inventory, 10)

        self.assertEqual(order.calculate_total(), 180)

    def test_stock_update(self):
        self.cart.add_item("P1", 2)
        order = Order(self.cart, self.inventory)

        order.place_order()

        self.assertEqual(
            self.inventory.data["P1"]["stock"], 3
        )

    def test_invalid_quantity(self):
        with self.assertRaises(ValueError):
            self.cart.add_item("P1", 0)


if __name__ == "__main__":
    unittest.main()
```
